### Model Integration with OpenAI, Google Gemini and GROQ

```python
import os;
from dotenv import load_dotenv;
load_dotenv();

os.environ["OPENAI_API_KEY"] = os.getenv("OPENAI_API_KEY");
os.environ["GOOGLE_API_KEY"] = os.getenv("GOOGLE_API_KEY");
os.environ["GROQ_API_KEY"] = os.getenv("GROQ_API_KEY");
```

```python
from langchain.chat_models import init_chat_model;
model = init_chat_model("gemini-3.5-flash-lite", model_provider="google_genai")
model
```

```python
## invoke the model
# res = model.invoke("What is the capital of France?");
res = model.invoke("Hello how are you?");
print(res.content);
```

### Streaming
Streaming is when the agent wants to show the output while processing the prompt. We do it using .stream

```python
for chunk in model.stream("Write me a 200 word essay on the importance of AI in modern society."):
    print(chunk.text)
```

### Batch
batch is when there are multiple requests and they're being processed paralley. It reduce cost and improve performance.

```python
responses = model.batch([
    "why is the sky blue?",
    "what is the capital of France?",
    "what is quantum computing?"
],
config = {
    'max_concurrency': 5 # Limit to the parallel calls
}
);

for response in responses:
    print(response)
```
