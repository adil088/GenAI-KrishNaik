### Messages
Messages are the fundamental unit of context for models in Langchain. They represent the input and output of the models, carrying both the content and metadata needed to represent the state of a conversation while interacting with LLM. Messages are object that contain:

1. Role - Identifies the message type (eg. system, user)
2. Content - Represents the actual content of the messages (like text, image, audio, documents, etc)
3. Metadata - Optional fields such as response information, message IDs, and token usage.

Langchain provides a standard message type that works accross all the model providers, ensuring consistent behaviour regardless of the model being called.
